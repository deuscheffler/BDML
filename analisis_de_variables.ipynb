```python
# ==========================
# LIBRERÍAS
# ==========================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ==========================
# CARGAR DATASET
# ==========================

df = pd.read_csv("DataCoSupplyChainDataset.csv", encoding='latin1')


df = df.rename(columns={
    'Type': 'Tipo_Pago',
    'Days for shipping (real)': 'Dias_Envio_Real',
    'Days for shipment (scheduled)': 'Dias_Envio_Programado',
    'Benefit per order': 'Beneficio_Por_Pedido',
    'Sales per customer': 'Ventas_Por_Cliente',
    'Delivery Status': 'Estado_Entrega',
    'Late_delivery_risk': 'Riesgo_Entrega_Tardia',
    'Category Id': 'ID_Categoria',
    'Category Name': 'Nombre_Categoria',
    'Customer City': 'Ciudad_Cliente',
    'Customer Country': 'Pais_Cliente',
    'Customer Email': 'Correo_Cliente',
    'Customer Fname': 'Nombre_Cliente',
    'Customer Id': 'ID_Cliente',
    'Customer Lname': 'Apellido_Cliente',
    'Customer Password': 'Contrasena_Cliente',
    'Customer Segment': 'Segmento_Cliente',
    'Customer State': 'Estado_Cliente',
    'Customer Street': 'Direccion_Cliente',
    'Customer Zipcode': 'Codigo_Postal_Cliente',
    'Department Id': 'ID_Departamento',
    'Department Name': 'Nombre_Departamento',
    'Latitude': 'Latitud',
    'Longitude': 'Longitud',
    'Market': 'Mercado',
    'Order City': 'Ciudad_Pedido',
    'Order Country': 'Pais_Pedido',
    'Order Customer Id': 'ID_Cliente_Pedido',
    'order date (DateOrders)': 'Fecha_Pedido',
    'Order Id': 'ID_Pedido',
    'Order Item Cardprod Id': 'ID_Producto_Item',
    'Order Item Discount': 'Descuento_Item',
    'Order Item Discount Rate': 'Tasa_Descuento_Item',
    'Order Item Id': 'ID_Item_Pedido',
    'Order Item Product Price': 'Precio_Producto_Item',
    'Order Item Profit Ratio': 'Ratio_Ganancia_Item',
    'Order Item Quantity': 'Cantidad_Item',
    'Sales': 'Ventas',
    'Order Item Total': 'Total_Item_Pedido',
    'Order Profit Per Order': 'Ganancia_Por_Pedido',
    'Order Region': 'Region_Pedido',
    'Order State': 'Estado_Pedido',
    'Order Status': 'Estado_Orden',
    'Order Zipcode': 'Codigo_Postal_Pedido',
    'Product Card Id': 'ID_Tarjeta_Producto',
    'Product Category Id': 'ID_Categoria_Producto',
    'Product Description': 'Descripcion_Producto',
    'Product Image': 'Imagen_Producto',
    'Product Name': 'Nombre_Producto',
    'Product Price': 'Precio_Producto',
    'Product Status': 'Estado_Producto',
    'shipping date (DateOrders)': 'Fecha_Envio',
    'Shipping Mode': 'Modo_Envio'
})

# Verificar nuevos nombres
for columna in df.columns:
    print(columna)
```

```python
resumen = pd.DataFrame({
    'Variable': df.columns,
    'Tipo': df.dtypes.values,
    'Nulos': df.isnull().sum().values,
    'Unicos': df.nunique().values
})

print(resumen)
```

```python
print("\nTotal de variables:", len(df.columns))

for i, columna in enumerate(df.columns, start=1):
    print(f"{i}. {columna}")
```

```python
columnas_modelo = [
    'Tipo_Pago',
    'Estado_Entrega',
    'Nombre_Categoria',
    'Segmento_Cliente',
    'Mercado',
    'Pais_Pedido',
    'Region_Pedido',
    'Estado_Orden',
    'Modo_Envio',
    'Dias_Envio_Real',
    'Dias_Envio_Programado',
    'Ventas',
    'Beneficio_Por_Pedido',
    'Ganancia_Por_Pedido',
    'Cantidad_Item',
    'Precio_Producto',
    'Descuento_Item',
    'Riesgo_Entrega_Tardia'
]
```

```python
# ==========================
# DATASET PARA ANÁLISIS
# ==========================

df_modelo = df[columnas_modelo].copy()

print("DIMENSIONES DEL DATASET")
print(df_modelo.shape)

print("\nPrimeras filas:")
print(df_modelo.head())
```

```python
# ==========================
# ESTADÍSTICAS DESCRIPTIVAS
# ==========================

print("ESTADÍSTICAS DESCRIPTIVAS")
print(df_modelo.describe())
```

```python
# Variables numéricas
numericas = [
    'Dias_Envio_Real',
    'Dias_Envio_Programado',
    'Ventas',
    'Beneficio_Por_Pedido',
    'Ganancia_Por_Pedido',
    'Cantidad_Item',
    'Precio_Producto',
    'Descuento_Item',
    'Riesgo_Entrega_Tardia'
]

fig, axes = plt.subplots(3,3, figsize=(16,12))

for ax, col in zip(axes.flatten(), numericas):

    sns.histplot(
        df_modelo[col],
        kde=True,
        ax=ax
    )

    ax.set_title(col)

plt.tight_layout()
plt.show()
```

```python
# ==========================
# VARIABLES CATEGÓRICAS
# ==========================

categoricas = [
    'Tipo_Pago',
    'Estado_Entrega',
    'Nombre_Categoria',
    'Segmento_Cliente',
    'Mercado',
    'Pais_Pedido',
    'Region_Pedido',
    'Estado_Orden',
    'Modo_Envio'
]

# VARIABLES CATEGÓRICAS
# ==========================

for col in categoricas:

    plt.figure(figsize=(10,5))

    df_modelo[col].value_counts().head(10).plot(
        kind='bar'
    )

    plt.title(f'Top categorías - {col}')
    plt.ylabel('Cantidad')
    plt.xticks(rotation=45)

    plt.show()
```

```python
# Matriz de correlación
plt.figure(figsize=(12,8))

sns.heatmap(
    df_modelo[numericas].corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Matriz de Correlación')

plt.show()
```

```python
# OUTLIERS
# ==========================

fig, axes = plt.subplots(3,3, figsize=(16,12))

for ax, col in zip(axes.flatten(), numericas):

    sns.boxplot(
        x=df_modelo[col],
        ax=ax
    )

    ax.set_title(col)

plt.tight_layout()
plt.show()
```

```python
# VARIABLE OBJETIVO
plt.figure(figsize=(6,4))

sns.countplot(
    data=df_modelo,
    x='Riesgo_Entrega_Tardia'
)

plt.title('Distribución Riesgo Entrega Tardía')

plt.show()
```

```python
# ENVÍO VS RIESGO
# ==========================

plt.figure(figsize=(8,5))

sns.boxplot(
    data=df_modelo,
    x='Riesgo_Entrega_Tardia',
    y='Dias_Envio_Real'
)

plt.title('Días de Envío Real vs Riesgo')

plt.show()
```

```python
# VENTAS POR PAÍS
# ==========================

ventas_pais = (
    df_modelo
    .groupby('Pais_Pedido')['Ventas']
    .sum()
    .sort_values(ascending=False)
    .head(15)
)

plt.figure(figsize=(12,5))

ventas_pais.plot(kind='bar')

plt.title('Top 15 Países por Ventas')
plt.ylabel('Ventas')

plt.show()
```

```python
# Estado de entrega
plt.figure(figsize=(8,5))

sns.countplot(
    y='Estado_Entrega',
    data=df,
    order=df['Estado_Entrega'].value_counts().index
)

plt.title('Estado de Entrega')
plt.show()
```

```python
# Ventas por mercado
ventas_mercado = (
    df.groupby('Mercado')['Ventas']
      .sum()
      .sort_values(ascending=False)
)

ventas_mercado.plot(
    kind='bar',
    figsize=(10,5)
)

plt.title('Ventas por Mercado')
plt.show()
```

```python
# TOP CATEGORÍAS POR VENTAS
# ==========================

top_categorias = (
    df
    .groupby('Nombre_Categoria')['Ventas']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12,5))

top_categorias.plot(kind='bar')

plt.title('Top Categorías por Ventas')
plt.ylabel('Ventas')

plt.show()
```

```python
# Riesgo por región
riesgo_region = (
    df_modelo
    .groupby('Region_Pedido')['Riesgo_Entrega_Tardia']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12,5))

riesgo_region.plot(kind='bar')

plt.title('Riesgo de Entrega Tardía por Región')
plt.ylabel('Promedio')

plt.show()
```

```python
# Riesgo por modo de envío
plt.figure(figsize=(10,5))

sns.barplot(
    data=df_modelo,
    x='Modo_Envio',
    y='Riesgo_Entrega_Tardia'
)

plt.title('Riesgo por Modo de Envío')

plt.xticks(rotation=45)

plt.show()
```

```python
# SEGMENTACIÓN DE CLIENTES
# ==========================

plt.figure(figsize=(10,5))

sns.countplot(
    data=df_modelo,
    y='Segmento_Cliente',
    order=df_modelo['Segmento_Cliente']
    .value_counts()
    .index
)

plt.title('Segmentos de Clientes')

plt.show()
```

```python
# ==========================
# VENTAS VS BENEFICIO
# ==========================

plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df_modelo,
    x='Ventas',
    y='Beneficio_Por_Pedido',
    alpha=0.5
)

plt.title('Ventas vs Beneficio')

plt.show()
```

```python
# Ventas por categoría
top_categoria = (
    df.groupby('Nombre_Categoria')['Ventas']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_categoria.plot(
    kind='bar',
    figsize=(12,5)
)

plt.title('Top Categorías por Ventas')
plt.show()
```

```python
# MERCADO VS MODO ENVÍO
# ==========================

tabla = pd.pivot_table(
    df_modelo,
    values='Riesgo_Entrega_Tardia',
    index='Mercado',
    columns='Modo_Envio',
    aggfunc='mean'
)

plt.figure(figsize=(12,6))

sns.heatmap(
    tabla,
    annot=True,
    cmap='Reds'
)

plt.title('Riesgo por Mercado y Modo de Envío')

plt.show()
```

```python
# DISTRIBUCIÓN GEOGRÁFICA
# ==========================

plt.figure(figsize=(12,6))

plt.scatter(
    df['Longitud'],
    df['Latitud'],
    alpha=0.2
)

plt.title('Distribución Geográfica de Pedidos')
plt.xlabel('Longitud')
plt.ylabel('Latitud')

plt.show()
```

```python
# RELACIONES
# ==========================

variables_pairplot = [
    'Ventas',
    'Cantidad_Item',
    'Precio_Producto',
    'Beneficio_Por_Pedido',
    'Dias_Envio_Real',
    'Riesgo_Entrega_Tardia'
]

sns.pairplot(
    df_modelo[variables_pairplot],
    hue='Riesgo_Entrega_Tardia'
)

plt.show()
```
